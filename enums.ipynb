<h2>Enums</h2>

```python
from enum import Enum, auto
```

```python
class Color(Enum):
    RED = 1
    GREEN = 2
    BLUE = 3
```

Loop over enum items as follows:

```python
for color in Color:
    print(f"{color.name}: {color.value}")
```

Use call syntax to get enum members by values.
Use index syntax to get enum members by their names.

```python
Color['RED'] == Color.RED == Color(1)
```

If the actual values don't have any significance in your current context, use <i>enum.auto()</i> instead and it will automatically assign values:

```python
class Days(Enum):
    MONDAY = auto()
    TUESDAY = auto()
```

```python
# Assigns integer values by default
Days.MONDAY.value, Days.TUESDAY.value
```

<h3>Advanced Uses of Enums</h3>

Try not to use any mutable values enum members. Even if enum will take anything, the enumeration values won't update.

```python
class ColorType:
    def __init__(self, name):
        self._name = name
        
class TypedEnum(Enum):
    GREEN = ColorType("green")
    RED = ColorType("red")
    
for member in TypedEnum:
    print(member, member.name, member.value._name)
```

```python
TypedEnum.RED._name = "another red" # Note -- this is no-op !!! 
for member in TypedEnum:
    print(member, member.name, member.value._name)
```

But if you really need it, at least use its own objects as values:

```python
class AdvancedColors(Enum):
    RED = ('red'),
    GREEN = ('green')
    
    
    def __init__(self, color_str):
        self.__color_str = color_str
        
    @property
    def color_str(self):
        return self.__color_str
    
    @color_str.setter
    def color_str(self, value):
        self.__color_str = value
```

```python
print(f"{AdvancedColors.RED}: value={AdvancedColors.RED.value}, color_str={AdvancedColors.RED.color_str}")
```

```python
AdvancedColors.RED.color_str = "light_red" # Will be accepted
print(f"{AdvancedColors.RED}: value={AdvancedColors.RED.value}, color_str={AdvancedColors.RED.color_str}")
```

<b>NOTE:</b> Starting from Python <i>3.11</i> there will be <i>IntEnum</i>, <i>StrEnum</i> etc. in the <i>enum</i> module. <i>3.11</i> also brings new enum anntoations such as <i>member</i>, <i>nonmember</i>, <i>property</i> etc. that allow more flexibility.
