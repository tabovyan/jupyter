## Match Statement

Match statement is like `switch` in Java. Only the first pattern that matches will get executed. It can extract components from the value into variables.

```python
def http_error(status):
    match status:
        case 400:
            return "Bad request"
        case 403 | 404:
            return "Not found"
        case _:
            return "Unexpected status"

print(http_error(404), ",", http_error(1000))
```

`match` can look like unpacking assignments where variables can be binded:

```python
# point is an (x,y) tuple
def match_point(point):
    match point:
        case (0,0):
            print("Origin")
        case (0, y):
            print(f"x is 0, y={y}")
        case (x, 0):
            print(f"x={x}, y is 0")
        case (x,y):
            print(f"x={x},y={y}")
        case _:
            print("Not an (x,y) tuple")
```

```python
match_point((0, 0))
match_point((0, 5))
match_point((5, 0))
match_point((5, 5))
match_point("blah")
```

`match` works with classes. In `case` statements specify the class you're expecting with arguments list resembling a constructor, but with variables that will bind the values:

```python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int

def where_is(point):
    match point:
        case Point(x=0, y=0):
            print("Origin")
        case Point(x=0, y=y):
            print(f"x is 0, y={y}")
        case Point(x=x, y=0):
            print(f"x={x}, y is 0")
        case Point():
            print("Somewhere else")
        case _:
            print("Not a point")
```

```python
where_is(Point(0,0))
where_is(Point(0,5))
where_is(Point(5,0))
where_is(Point(5,5))
where_is((1,1))
```

You can even use `if` clause to a pattern (wow!). This is called *guard*

```python
point = Point(3,3)

match point:
    case Point(x,y) if x == y:
        print(f"x==y at {x}")
    case Point(x,y):
        print("x <> y")
```

The most natural use will be with Enums of course:

```python
from enum import Enum
class Color(Enum):
    RED = 'red'
    GREEN = 'green'
    BLUE = 'blue'

color = Color(input("Enter your choice of 'red', 'blue' or 'green': "))

match color:
    case Color.RED:
        print("I see red!")
    case Color.GREEN:
        print("Grass is green")
    case Color.BLUE:
        print("I'm feeling the blues :(")
```

Use `match` with lists

```python
def match_list(my_list):
    match my_list:
        case []:
            print("Empty list")
        case [Point(0,0)]:
            print("Single element list with Origin")
        case [Point(0, y), Point(x, 0)]:
            print("2 elements: One with y == 0 another with x == 0")
        case [Point(0, y1), Point(0, y2), Point(0, y3)]:
            print(f"3 elements: all with x == 0 and y=({y1}, {y2}, {y3})")
        case "blah":
            print("blah string")
        case _:
            print("Something else")
            
```

```python
match_list([])
match_list([Point(0,1), Point(1,0)]) 
match_list([Point(1,0), Point(0,1)]) # Same as above, order flipped
match_list([Point(0, 1), Point(0,10), Point(0,100)])
match_list("blah")
match_list([1,2])
```

You get the idea. Mix and match all you like, unpack all you like, make it even more versatile with `if` guard expressions.
