<h2>The Basics</h2>

Numpy's main object is a homogeneous multidimensional array - all elements are of the same type. 

```python
import numpy as np
```

```python
# Reshape to 3 rows and 5 columns
# Or, 5 in the inner (last) dimension, 3 on the outer (first) dimension
a = np.arange(15).reshape(3, 5)
a
```

```python
# properties of ndarray
print(f"type: {type(a)}, shape: {a.shape}, ndim: {a.ndim}, dtype: {a.dtype}, dtype_name: {a.dtype.name}")
```

<h3>Create Arrays</h3>

```python
a = np.array([1,2,3]) # from list
b = np.array([(1.0,2,3), (4,5,6)]) # from list of tuples
c = np.array([1,2], dtype=complex) # specify dtype explicitly
```

```python
b
```

```python
a.dtype, b.dtype, c.dtype
```

```python
d = np.zeros((3,4)) # 3x4 filled with 0s
d
```

```python
e = np.ones((2,3,4), dtype=np.int16)
e
```

```python
f = np.empty((2,3)) # may vary
f
```

```python
g = np.arange(10,30, 5) # interval with step
g
```

```python
# can't know where it will finish and how many items it will end up with? use linspace
h = np.arange(0 ,2, 0.3) 
h
```

```python
i = np.linspace(0,2,9) # how many points
i
```

<h4>Random generation</h4>

```python
# Samples from standard normal (z) distribution
# Note: dimensions are passed directly and not as tuple
np.random.randn(2, 3)
```

<h2>Basic Operations</h2>

All arithmetic operations apply element-wise.

```python
a = np.array([1,2,3,4])
b = np.array([1,1,1,1])
```

```python
a + b
```

```python
a ** (b + 1)
```

```python
np.sin(a)
```

```python
a = np.array([[1,1], [0,1]])
b = np.array([[2,0], [3,4]])
```

```python
a
```

```python
b
```

```python
a * b # this is not a matrix multiplication
```

```python
a @ b # this is a matrix multiplication
```

```python
a.dot(b) # and so is this
```

<h3>Modify the array</h3>

```python
a += 1
a
```

```python
b = np.array([1.9, 2.9, 3.9, 4.9])
```
