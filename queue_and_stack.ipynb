```python
stack2.pop()
```

```python
from queue import LifoQueue, Empty
```

```python
stack3 = LifoQueue()
```

```python
# DO NOT USE THIS CONSTRUCT!
if stack3:
    print(f"stack2 is not empty {stack3}")
```

```python
# USE THIS INSTEAD
if stack3.qsize():
    print(f"stack2 is not empty {stack3}")
```

```python
stack3.put(1)
stack3.put(2)
```

```python
stack3.get()
```

```python
stack3.get()
```

```python
# Will hang for 5 sec, then raise Empty error
try:
    stack3.get(timeout=5)
except Empty:
    print("Stack is empty")
```

<h3>Synchronized queue</h3>

```python
from queue import Queue, Empty
```

```python
my_queue = Queue()
```

```python
my_queue.put(1)
my_queue.put(2)
```

```python
my_queue
```

```python
my_queue.qsize() == 2
```

```python
my_queue.get_nowait()
```

```python
my_queue.get()
```

```python
try:
    my_queue.get_nowait()
except Empty:
    print("no more elems left")
```

```python
try:
    await my_queue.get(timeout=2)
except Empty:
    print("no more elems left after waiting for 2 sec")
```

<h2>Prioritity Queue</h2>

```python
pqueue = []
```

```python
from heapq import heappop, heappush, heapify
```

```python
heappush(pqueue, 1)
heappush(pqueue, 10)
heappush(pqueue, -3)
heappush(pqueue, -10)
```

```python
heappop(pqueue)
```

```python
heappop(pqueue)
```

<h3>Synchronized PQueue</h3>

```python
from queue import PriorityQueue
```

```python
pqueue = PriorityQueue()
```

```python
pqueue.put(1)
pqueue.put(10)
pqueue.put(-3)
pqueue.put(-10)
```

```python
pqueue.get()
```

```python
pqueue.get_nowait()
```

```python
# block at most 2 seconds
pqueue.get(block=True, timeout=2)
```
