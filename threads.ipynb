```python
import threading
import time
```

```python
def print_numbers():
    for i in range(10):
        print(f"number {i}")
        time.sleep(1)

def print_letters():
    for letter in "abcdefghi":
        print(f"letter {letter}")
        time.sleep(1)
```

```python
# Create thread objects
thread1 = threading.Thread(target=print_numbers)
thread2 = threading.Thread(target=print_letters)

# start the threads
thread1.start()
thread2.start()

print("This line will definitely be printed before threads are finished")
```

```python
# Create thread objects again, but this time cell will wait for them to finish
thread1 = threading.Thread(target=print_numbers)
thread2 = threading.Thread(target=print_letters)

# start the threads
thread1.start()
thread2.start()

thread1.join()
thread2.join()

print("This line will definitely be printed after threads are finished")
```

<h2>Using Locks</h2>

```python
counter = 0
my_lock = threading.Lock()
```

```python
def increment():
    global counter
    for _ in range(1000):
        with my_lock:
            counter += 1
```

```python
threads = [threading.Thread(target=increment) for _ in range(10)]
```

```python
for thread in threads:
    thread.start()
    
for thread in threads:
    thread.join()
    
print(f"counter={counter}")
```

<h2>Using thread executors</h2>

```python
import concurrent.futures
```

```python
def task(n):
    print(f"task {n} starting")
    time.sleep(1)
    print(f"task {n} completed")
    return f"result of task {n}"
```

```python
with concurrent.futures.ThreadPoolExecutor(max_workers=3) as executor:
    # submit tasks to the executor
    futures = [executor.submit(task, i) for i in range(5)]
    
    for future in concurrent.futures.as_completed(futures):
        print(future.result())

print("All tasks finished execution now")
```
