<h2>Recursion with Open and Close Parenthesis</h2>

```python
import re
def get_max_length_of_valid_subsequence(word: str) -> int:
    stack: list[int] = [-1]
    current_max = 0
    word = re.sub("[^\(\)]+", "", word) # remove all non-paren characters
    print(f"Finding the longest valid sequence of ( and ) characters in {word}")
    for i in range(len(word)):
        if word[i] == "(":
            stack.append(i)
        else:
            stack.pop()
            if len(stack) == 0:
                stack.append(i)
            start = stack[-1]
            if start < i:
                # Found match from a position after last valid subsequence finished to current ')' s position
                print(f"  Encountered valid sequence: {word[stack[-1]+1:i+1]}")
                current_max = max(current_max, i - start)    
    print(f"  The longest valid subsequence is of length {current_max}")
    return current_max
```

```python
get_max_length_of_valid_subsequence("((()))")
```

```python
get_max_length_of_valid_subsequence("))()()()()))")
```

<h2>Redefine + operator</h2>

```python
class Expenses:
    def __init__(self, rent, groceries):
        self.rent = rent
        self.groceries = groceries

    def __add__(self, other):
        return Expenses(self.rent + other.rent, 
                        self.groceries + other.groceries)
    
    def __sub__(self, other):
        return Expenses(self.rent - other.rent, 
                        self.groceries - other.groceries)
    
    def __repr__(self):
        # custom printable version of the object
        return f"(rent={self.rent}, groceries={self.groceries})"
```

```python
april = Expenses(1000, 200)
may = Expenses(1000, 300)
added = april + may
subtracted = may - april
print(added, subtracted)
```

<h2>copy() vs deepcopy()</h2>

`shallow copy` constructs a new compound object, but inserts references into it to the objects found in orginal.

`deep copy` constructs a new compound object, but recursively inserts copies into it of the objects found in the original.

```python
first_list = [[1,2], ["a", "b", "c"]]
second_list = first_list.copy()
```

```python
second_list[0][1] = 10
```

```python
# Both lists point to the same inner sublist
print(first_list)
print(second_list)
```

```python
import copy
first_list = [[1,2], ["a", "b", "c"]]
second_list = copy.deepcopy(first_list)
```

```python
second_list[0][1] = 10
```

```python
# Different sublists for both
print(first_list)
print(second_list)
```
