<h3>Regex</h3>

```python
# Cheat sheet:
"""
re.fullmatch(pattern, s) -> re.Match # full match
re.match(pattern, s)  -> re.Match # from the beginning
re.search(pattern, s)  -> re.Match # anywhere in the string
re.findall(pattern, s)  -> list[str] # all matches
re.finditer(pattern, s)  -> iterable[re.Match] # all matches

re.Match has: m.group(), m.start(), m.end()

"""
import re
```

```python
# full match
s = "1234d"
if match1 := re.fullmatch("\d+", s):
    print("fully matches")
    
# match from a beginning of string
if match2 := re.match("\d+", s):
    print(f"there is a match at the beginning: {match2.group()}")
```

```python
match2.group(), match2.start(), match2.end()
```

```python
# for complex patterns, create a pattern first
p = re.compile(r"[0-9]+")
```

```python
# match from any location
re.search(p, "abc123d456")
```

```python
# same result, but using the same re methods on a pattarn
p.search("abc123d456")
```

```python
type(re.findall(p, "abc123d456"))
```

```python
# find all matches
for sub in re.findall(p, "abc123d456"):
    print(sub, type(sub))
```

```python
# for locations of matches, use iterator
for m in re.finditer(p, "abc123e456"):
    print(f"{m.group()}: {m.span()}")
```

<h3>Magic Methods</h3>

```python
# Magic methods
class MyNumber:
    def __init__(self, num: int):
        self.num = num
    
    def __add__(self, num_obj):
        return MyNumber(self.num + num_obj.num)
    
    def __sub__(self, num_obj):
        return MyNumber(self.num - num_obj.num)
    
    def __mul__(self, num_obj):
        return MyNumber(self.num * num_obj.num)
    
    def __truediv__(self, num_obj):
        return MyNumber(self.num / num_obj.num)
    
    def __floordiv__(self, num_obj):
        return MyNumber(self.num // num_obj.num)
    
    def __pow__(self, num_obj):
        return MyNumber(self.num ** num_obj.num)
    
    def __mod__(self, num_obj):
        return MyNumber(self.num % num_obj.num)
    
    # Also add right hand methods, or __r*__() to be able to do e.g. 2 + obj when __radd__ is called
    # Also add i methods, or __i*__(self, other) to be able to do e.g. += 
    def __repr__(self):
        return str(self.num)
    
    
```

        self._index = 0
        return self
    
        # might as well just do the following and ignore __next__

        if len(self.items) == 0:
            raise StopIteration

            self._index = 0
            
        self._count += 1
        if self._count == MyCollection.MAX_ITERATION:

```python
class Node:
    def __init__(self, name: str):
        self.name = name
        self.neighbors = {} # name->weight/dist dict
        
        # may not need all of them
        self.visited = False
        self.dist = float("inf")
        self.prev = None
        
    def add_neighbor(self, name: str, dist: float):
        if name in self.neighbors:
            raise ValueError(f"Node {name} is already in the neighbors list of {self.name}")
        self.neighbors[name] = dist
        
    def reset(self):
        self.visited = False
        self.dist = float("inf")
        self.prev = None

class Graph:
    def __init__(self):
        self.nodes = {}
        self.edges = []
    
    def add_node(self, name: str):
        if name not in self.nodes:
            node = Node(name)
            self.nodes[name] = node
        
    def add_edge(self, from_node: str, to_node: str, dist: float):
        if from_node not in self.nodes:
            self.nodes[from_node] = Node(from_node)
        if to_node not in self.nodes:
            self.nodes[to_node] = Node(to_node)
        
        self.nodes[from_node].add_neighbor(to_node, dist)
        self.edges.append((dist, from_node, to_node))
        
class UndirectedGraph(Graph):
    def __init__(self):
        self.nodes = {}
        self.edges = []
    
    def add_node(self, name: str):
        if name not in self.nodes:
            node = Node(name)
            self.nodes[name] = node
            
    def get_node(self, name: str):
        return self.nodes[name] if name in self.nodes else None
        
    def add_edge(self, from_node: str, to_node: str, dist: float):
        if from_node not in self.nodes:
            self.nodes[from_node] = Node(from_node)
        if to_node not in self.nodes:
            self.nodex[to_node] = Node(to_node)
        
        # only diff is, neighbor relation is both ways
        self.nodes[from_node].add_neighbor(to_node, dist)
        self.nodes[to_node].add_neighbor(from_node, dist)
        
        self.edges.append((dist, from_node, to_node))
    
```

```python
# Find shortest path from node1 to node2, and if it doesn't exist, return -1 
# All distances have to be positive!!! - Dijkstra's algorithm requires this, otherwise it won't work
```

```python
import heapq
```

```python
# Can work the same way for undirected graph
def find_shortest_path(from_node: Node, to_node: Node, graph: DirectedGraph):
    candidates = []
    visited = set()
    from_node.dist = 0
    from_node.visited = True
    
    heapq.heappush(candidates, (0, from_node.name))
    
    while len(candidates) > 0:
        next_node_name = heapq.heappop(candidates)[1]
        next_node = graph.get_node(next_node_name)
        next_node.visited = True
        
        if next_node_name == to_node.name:
            # walk backwards and construct the path
            n = next_node
            path = [n.name]
            while n.prev is not None:
                n = graph.get_node(n.prev)
                path.append(n.name)
            return (next_node.dist, path[::-1])
        
        for neighbor, dist in next_node.neighbors.items():
            neighbor_node = graph.get_node(neighbor)
            
            if not neighbor_node.visited:
                if neighbor_node.dist > next_node.dist + dist:
                    neighbor_node.dist = next_node.dist + dist
                    neighbor_node.prev = next_node.name
                    
                # NOTE: Its possible that same node can go to heap with diff distances, but who cares as smallest distance will come out first
                heapq.heappush(candidates, (neighbor_node.dist, neighbor_node))
    return (-1, "UNREACHABLE")
```

<h3>Find shortest path from [0,0] -> [m-1,n-1] on matrix with obstacles</h3>

```python
import heapq
```

```python
# Note: its possible that the same (i,j) cell will go into the PQueue with diff distances, but who cares as the smallest dist will be visited first
def find_shortest_path(matrix: list[list[int]]) -> list[tuple[int]]:
    m, n = len(matrix), len(matrix[0])
    visited = set() # Dijksta's visited cells that have min distance determined
    items = [] # This is the priority queue that helps pick the next closest cell. Its items are: (distance,i,j,prev_cell) tuples
    
    distances = dict() # (i,j) -> distance, prev mapping to quickly fetch [0,0]->[i,j] distance so far
    distances[(0, 0)] = (0, None)
    results = [] # resulting path
    
    start = (0, 0, 0, None)
    heapq.heappush(items, start)
    while len(items) > 0:
        cur_dist, i, j, prev = heapq.heappop(items)
        if i == m - 1 and j == n - 1:
            # Reached. Construct the path and return.
            i_, j_ = m - 1, n - 1
            results.append((i_, j_))
            while distances[(i_, j_)][1] is not None:
                i_, j_ = distances[(i_, j_)][1] # prev cell
                results.append((i_, j_))
                
            return results[::-1]
        
        # (i,j) is the next closest node, mark as visited
        visited.add((i, j))
        
        # Considering i - 1, j cell
        if i > 0 and matrix[i-1][j] != 1 and (i - 1, j) not in visited:
            if (i - 1, j) not in distances or distances[(i -1, j)][0] > cur_dist + 1:
                distances[(i - 1, j)] = (cur_dist + 1, (i,j))
                heapq.heappush(items, (cur_dist + 1, i - 1, j, (i, j)))
        
        # Considering i, j - 1 cell
        if j > 0 and matrix[i][j - 1] != 1 and (i, j - 1) not in visited:
            if (i, j - 1) not in distances or distances[(i, j - 1)][0] > cur_dist + 1:
                distances[(i, j - 1)] = (cur_dist + 1, (i,j))
                heapq.heappush(items, (cur_dist + 1, i, j - 1, (i, j)))
        
        # Considering i + 1, j cell
        if i < m - 1 and matrix[i+1][j] != 1 and (i + 1, j) not in visited:
            if (i + 1, j) not in distances or distances[(i + 1, j)][0] > cur_dist + 1:
                distances[(i + 1, j)] = (cur_dist + 1, (i,j))
                heapq.heappush(items, (cur_dist + 1, i + 1, j, (i, j)))
                
        # Considering i, j + 1 cell
        if j < n - 1 and matrix[i][j + 1] != 1 and (i, j + 1) not in visited:
            if (i, j + 1) not in distances or distances[(i, j + 1)][0] > cur_dist + 1:
                distances[(i, j + 1)] = (cur_dist + 1, (i,j))
                heapq.heappush(items, (cur_dist + 1, i, j + 1, (i, j)))
    return []
    
```

```python
matrix = [[0,1,0], [0, 1, 0], [0, 0, 0]]
```

```python
for t in find_shortest_path(matrix):
    print(f"-> {t}", end = " ")
```

<h3>Topological Sort</h3>

<h2>Data Structures</h2>

Did you know that you can implement a stack that is constant time, but also arbitrary removals by value are possible?
Well, the stack can be a Linked List and each value can point to a node in hashtable. Now, every time you pop, you pop from the end. Every time you remove,
you link previous and next items together.

```python
class Node:
    """Linked list node definition."""
    def __init__(self, val: str):
        self.val = val
        self.prev_node = None
        self.next_node = None
        
class MyStack:
    def __init__(self):
        self.head = None
        self.items = dict()
    
    def push(self, val: str):
        # delete in case if it exists
        self.delete(val)
        
        # create a fresh new node
        new_node = Node(val)
        self.items[val] = new_node
        
        # adjust head.next and new_node.prev 
        if self.head is not None:
            self.head.next_node = new_node
        new_node.prev_node = self.head
        
        # we have a new head
        self.head = new_node
    
    def pop(self):
        if self.head is not None:
            val = self.head.val
            del self.items[val]
            
            # head changes, prev node's next becomes None
            self.head = self.head.prev_node
            if self.head is not None:
                self.head.next_node = None
                
            return val
        raise ValueError("Stack is empty")
            
    def delete(self, val: str):
        if val in self.items:
            node = self.items[val]
            prev_node = node.prev_node
            next_node = node.next_node
            
            # remove the node, make prev and next adjustments
            if prev_node is not None:
                prev_node.next_node = next_node
            if next_node is not None:
                next_node.prev_node = prev_node
            
            # in case if the deleted node is the head, change the head to its previous node
            if self.head.val == node.val:
                self.head = node.prev_node
                
    def empty(self):
        return self.head == None
    
    def __repr__(self):
        cur = self.head
        result_str = ""
        while cur is not None:
            result_str = result_str + " -> " + cur.val
            cur = cur.prev_node
        return result_str
            
```

```python
my_stack = MyStack()
my_stack.push("one")
my_stack.push("two")
my_stack.push("three")
my_stack.push("four")
my_stack.push("five")
print(my_stack)
my_stack.pop()
print(my_stack)
my_stack.delete("two")
print(my_stack)
my_stack.delete("four")
print(my_stack)
my_stack.pop()
print(my_stack)
my_stack.pop()
print(my_stack)
print(my_stack.empty())
```
