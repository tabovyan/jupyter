```python
from collections import Counter
```

```python
s = "abababac"
counter = Counter(s)
counter
```

```python
# 2 of the most common items
two_most_common: list[int] = counter.most_common(n=2)
for value,count in two_most_common:
    print(f"{value}: {count}")
```

```python
# sorted by most common to least common -- returns all items
counter.most_common()
```

```python
# total counts across all elements
counter.total()
```

```python
# add elem
counter.update("a")
counter
```

```python
# remove elem
counter.subtract("b")
counter
```

<h2>Counter as a Set</h2>

```python
counter2 = Counter("bbc")
```

```python
counter == counter2, counter <= counter2, counter >= counter2
```

```python
counter2.update("c")
counter2.update("d")
counter2
```

```python
# One contains another
counter2 < counter, counter2 > counter
```

```python
# set difference
counter2 - counter
```

```python
# set intersection
counter2 & counter
```

```python
# Set union
counter2 | counter
```
